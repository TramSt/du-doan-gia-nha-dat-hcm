## 4. Lọc và suy đoán quận/huyện.

sử dụng thư viện re, để trích xuất các quanba

```python
import re
import pandas as pd


def lam_sach_quan_huyen_tu_mo_ta(df):
    """
    Trích xuất Quận/Huyện từ cột Mo_ta_Dac_diem.

    Hỗ trợ:
        Quận 1 -> Quận 12
        Quận Nhà Bè
        Gò Vấp
        Tân Bình
        Bình Tân
        Bình Thạnh
        Hóc Môn
        Tân Phú
        Thủ Đức

    Không xử lý Phường/Xã.

    Chỉ điền vào df['quan'] nếu giá trị hiện tại đang thiếu.
    """

    df = df.copy()

    if "quan_huyen" not in df.columns:
        df["quan_huyen"] = pd.NA

    if "mo_ta_dac_diem" not in df.columns:
        raise KeyError("Không tìm thấy cột 'Mo_ta_Dac_diem'")

    def trich_quan(text):
        if pd.isna(text):
            return None

        text = str(text)

        # Chuẩn hóa khoảng trắng
        text = re.sub(r"\s+", " ", text).strip()

        # Chuyển về chữ thường để tìm kiếm
        text_lower = text.lower()

        # --------------------------------------------------
        # QUẬN 1 -> QUẬN 12
        # --------------------------------------------------
        match = re.search(
            r"\bquận\s*(1[0-2]|[1-9])\b",
            text_lower
        )

        if match:
            return f"Quận {match.group(1)}"

        # Hỗ trợ dạng Q.1 / Q1 / Q. 1
        match = re.search(
            r"\bq\.?\s*(1[0-2]|[1-9])\b",
            text_lower
        )

        if match:
            return f"Quận {match.group(1)}"

        # --------------------------------------------------
        # CÁC QUẬN/HUYỆN THEO TÊN
        # --------------------------------------------------

        danh_sach = [
            ("bình thạnh", "Quận Bình Thạnh"),
            ("gò vấp", "Quận Gò Vấp"),
            ("tân bình", "Quận Tân Bình"),
            ("bình tân", "Quận Bình Tân"),
            ("tân phú", "Quận Tân Phú"),
            ("nhà bè", "Huyện Nhà Bè"),
            ("hóc môn", "Huyện Hóc Môn"),
            ("thủ đức", "TP. Thủ Đức"),
        ]

        for tu_khoa, ket_qua in danh_sach:
            if tu_khoa in text_lower:
                return ket_qua

        return None

    # Trích xuất
    df["_quan_trich_xuat"] = df["mo_ta_dac_diem"].apply(trich_quan)

    # Chỉ điền vào dòng đang thiếu
    mask = (
        df["quan_huyen"].isna()
        | df["quan_huyen"].astype(str).str.strip().eq("")
    )

    df.loc[mask, "quan_huyen"] = df.loc[mask, "_quan_trich_xuat"]

    # Xóa cột tạm
    df.drop(columns=["_quan_trich_xuat"], inplace=True)

    return df
df_sach=lam_sach_quan_huyen_tu_mo_ta(df_sach)
df_sach
```

```python
import pandas as pd

def loc_outlier_iqr(df, columns=['gia_ban', 'dien_tich_m2']):
    """
    Lọc bỏ các dòng chứa outlier dựa trên phương pháp IQR.
    """
    for col in columns:
        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)
        IQR = Q3 - Q1
        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR

        df_filtered = df[
            (df[col].isna()) | 
            ((df[col] >= lower_bound) & (df[col] <= upper_bound))
        ]
    return df_filtered
df_sach = loc_outlier_iqr(df_sach, columns=['gia_ban', 'dien_tich_m2'])
df_sach
```

**Cần trả lời được:**
1. Đã loại bao nhiêu dòng ở mỗi bước? Vì sao?<br>
Ban đầu: 29.532 dòng × 11 cột.
Kiểm tra chất lượng dữ liệu:
Giá bán ≤ 0: 0 dòng
Diện tích < 0: 0 dòng

→ Không loại bỏ dòng nào ở bước này.
Lọc và trích xuất quận từ mô tả: không làm thay đổi số lượng dòng.
Lọc outlier: từ 29.532 dòng xuống còn 27.204 dòng, tức đã loại 2.328 dòng.
Mục đích của bước lọc outlier là loại bỏ những tin có giá bán hoặc diện tích quá bất thường, có thể gây ảnh hưởng đến kết quả phân tích/mô hình.
2. Ngưỡng lọc outlier chọn thế nào? Căn cứ vào đâu?

Sử dụng phương pháp IQR (Interquartile Range). Với mỗi biến cần kiểm tra outlier, tính:

$$ IQR = Q_3 - Q_1 $$

Trong đó:

\(Q_1\): phân vị thứ 25%
\(Q_3\): phân vị thứ 75%

Sau đó xác định ngưỡng:

$$ LowerBound = Q_1 - 1.5 \times IQR $$ $$ UpperBound = Q_3 + 1.5 \times IQR $$

Một giá trị được xem là outlier nếu:

$$ x < Q_1 - 1.5IQR $$

hoặc

$$ x > Q_3 + 1.5IQR $$

Các giá trị nằm trong khoảng:

$$ [Q_1 - 1.5IQR,\ Q_3 + 1.5IQR] $$

được giữ lại.

Căn cứ chọn ngưỡng 1.5 × IQR: đây là quy tắc Tukey phổ biến để xác định các giá trị bất thường dựa trên phân phối của chính dữ liệu, không phụ thuộc vào giả định dữ liệu phải phân phối chuẩn. Đồng thời, trong dữ liệu bất động sản, giá và diện tích thường có phân phối lệch nên IQR phù hợp hơn việc chỉ dùng trung bình và độ lệch chuẩn.

Lưu ý: các giá trị NaN được giữ lại trong bước lọc outlier, vì NaN không thể được kết luận là outlier chỉ dựa trên IQR.
